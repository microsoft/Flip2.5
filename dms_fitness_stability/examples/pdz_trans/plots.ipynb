# PDZ trans stability plots

These figures use only the saved `metadata.json`, `summary.csv`, and `overlaps.csv` files. They do not load `stability.nc` or reopen either HMC fit.

```python
# ruff: noqa: E402
# pyright: reportMissingImports=false
import sys
from pathlib import Path


def find_project_root() -> Path:
    for base in (Path.cwd(), *Path.cwd().parents):
        for candidate in (base, base / "dms-fitness-stability"):
            if (candidate / "src" / "dms_fitness_stability").is_dir():
                return candidate
    raise FileNotFoundError("Could not find the dms-fitness-stability repository.")


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT / "src") not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

from dms_fitness_stability import (
    plot_inclusion,
    plot_rank_intervals,
    plot_report,
)

RESULT_DIR = PROJECT_ROOT / "examples" / "pdz_trans" / "results"
RESULT_DIR
```

## Trans-1

A single call generates inclusion, overlap, reported-rank, proxy-value, and rank-calibration figures for the selected score.

```python
trans_1_figures = plot_report(
    RESULT_DIR / "trans-1",
    k=50,
    scores=("log_mean_growth",),
)
```

## Trans-2

The same command applies to another landscape by changing only the result directory.

```python
trans_2_figures = plot_report(
    RESULT_DIR / "trans-2",
    k=50,
    scores=("log_mean_growth",),
)
```

## Individual figures

Call an individual function when only one view is needed. Omit `scores` from `plot_report` to generate the full report for every saved score.

```python
inclusion_plot = plot_inclusion(
    RESULT_DIR / "trans-1",
    score="count_ratio",
    k=50,
)

rank_plot = plot_rank_intervals(
    RESULT_DIR / "trans-1",
    score="count_ratio",
    k=50,
)
```
